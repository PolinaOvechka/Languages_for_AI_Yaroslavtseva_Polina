# Работа с библиотекой Pandas

1. Загрузите набор данных из файла **car_data.csv** по примеру с лекции



```python
import pandas as pd
import numpy as np

df = pd.read_csv("car_data.csv")

print("Размер таблицы:", df.shape)
display(df[["Company", "Model", "Body Style", "Price ($)", "Annual Income"]].head())
```

2. Проведите анализ пропусков в данных. Удалите строки или столбцы, содержащие пропуски.

```python
print("Количество пропусков в каждом столбце:")
display(df.isna().sum())

rows_before = len(df)
df = df.dropna().reset_index(drop=True)

print("Удалено строк:", rows_before - len(df))
print("Осталось строк:", len(df))
```

3. Вычислите средний доход клиентов, купивших автомобиль марки Toyota Camry.

```python
camry = df[(df["Company"] == "Toyota") & (df["Model"] == "Camry")]
mean_income = camry["Annual Income"].mean()

print("Количество покупок Toyota Camry:", len(camry))
print("Средний годовой доход покупателей:", round(mean_income, 2))
```

4. Найдите топ 10 самых дорогих марок автомобилей за представленный период времени.


```python
top_companies = df.groupby("Company")["Price ($)"].max().nlargest(10)

display(top_companies.to_frame("Максимальная цена ($)"))
```

5. Найдите среднюю стоимость автомобилей по типам кузова (Body Style), используя метод groupby.

```python
body_prices = df.groupby("Body Style")["Price ($)"].mean().round(2)

display(body_prices.to_frame("Средняя цена ($)"))
```

6. Какие автомобили приобрели 10 клиентов с самым высоким доходом?

```python
columns = ["Company", "Model", "Price ($)", "Annual Income"]

high_income = df.nlargest(10, "Annual Income")
display(high_income[columns].reset_index(drop=True))
```

7. Какие автомобили приобрели 10 клиентов с самым низким доходом?


```python
low_income = df.nsmallest(10, "Annual Income")
display(low_income[columns].reset_index(drop=True))
```

8. Создайте новый столбец, значения которого равны логарифму от величины дохода клиентов.


```python
df["Log Income"] = np.log(df["Annual Income"])

display(df[["Annual Income", "Log Income"]].head())
```
