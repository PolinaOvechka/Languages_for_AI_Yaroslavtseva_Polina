```python
def task_10():
    filename = "example.txt"

    def add_text_to_file(filename, text):
        with open(filename, 'a') as file:
            file.write(text + '\n')

    def show_file_content(filename):
        print(f"\nСодержимое файла '{filename}':\n")
        with open(filename, 'r') as file:
            content = file.read()
            print(content)

    def read_last_n_lines(filename,n):
      with open(filename, 'r') as file:
          all_lines = file.readlines()
          last_lines = all_lines[-n:]
      return last_lines

    ### ТЕСТИРОВАНИЕ ###
    with open (filename, 'w') as file:
      pass
    for i in range (1, 11):
      add_text_to_file(filename, f'Строка номер {i}')

    show_file_content(filename)

    print('\nПоследние 3 строки:')
    last_3 = read_last_n_lines(filename, 3)
    for line in last_3:
      print(line.strip())


```

```python
def task_11():
    import pickle

    my_dict = {
        'имя': 'Полина',
        'возраст': 25,
        'образование': 'ТвГТУ'
    }

    print('Исходный словарь:')
    print(my_dict)

    filename = 'data.pkl'

    with open(filename, 'wb') as file:
        pickle.dump(my_dict, file)

    print(f"\nСловарь сохранён в файл '{filename}'")

    with open(filename, 'rb') as file:
        loaded_dict = pickle.load(file)

    print('\nПрочитанный словарь:')
    print(loaded_dict)

    print(f"\nСловари совпадают? {my_dict == loaded_dict}")
```

```python
def task_12():
    import json

    my_dict = {
        'имя': 'Полина',
        'возраст': 25,
        'образование': 'ТвГТУ'
    }

    print('Исходный словарь:')
    print(my_dict)

    filename = 'data.json'

    with open(filename, 'w') as file:
        json.dump(my_dict, file)

    print(f"\nСловарь сохранён в файл '{filename}'")

    with open(filename, 'r') as file:
        loaded_dict = json.load(file)

    print('\nПрочитанный словарь:')
    print(loaded_dict)

    print(f"\nСловари совпадают? {my_dict == loaded_dict}")
```
