## Гипотезы

### 1. Связь массы автомобиля и топливной экономичности

**Предположение:** чем тяжелее автомобиль, тем он менее экономичен

Исследуемые переменные:
- weight
- mpg

**Гипотезы:**


H0: корреляция между массой автомобиля и топливной экономичностью отсутствует (p = 0)


H1: существует отрицательная корреляция между массой автомобиля и топливной экономичностью (p < 0)


### 2. Различие топливной экономичности американских и японских автомобилей

**Предположение:** средняя топливная экономичность автомобилей из США и Японии различается

Исследуемые переменные:
- origin
- mpg

**Гипотезы:**


H0: средняя топливная экономичность американских и японских автомобилей одинакова (μ_USA = μ_Japan)


H1: средняя топливная экономичность американских и японских автомобилей различается (μ_USA ≠ μ_Japan)


```python
import matplotlib.pyplot as plt
from scipy import stats

alpha = 0.05

'Гипотеза 1'
sns.scatterplot(data=df, x="weight", y="mpg")
plt.title("Зависимость экономичности от массы автомобиля")
plt.xlabel("Масса автомобиля")
plt.ylabel("Топливная экономичность")
plt.show()

correlation, p_value = stats.spearmanr(
    df["weight"], df["mpg"], alternative="less"
)

print("Коэффициент корреляции Спирмена:", correlation)
print("p-value:", p_value)

if p_value <= alpha:
    print("Нулевая гипотеза отвергается")
else:
    print("Нет оснований отвергать нулевую гипотезу")
```

**Вывод:**

Между массой автомобиля и его топливной экономичностью существует сильная отрицательная связь. Это означает, что чем тяжелее автомобиль, тем обычно больше топлива он расходует на одинаковое расстояние. Т.к p-value меньше 0.05, мы отвергаем нулевую гипотезу в пользу альтернативной

```python
import matplotlib.pyplot as plt
from scipy import stats

alpha = 0.05

'Гипотеза 2'

mpg_usa = df[df["origin"] == "usa"]["mpg"]
mpg_japan = df[df["origin"] == "japan"]["mpg"]

print("Количество американских автомобилей:", len(mpg_usa))
print("Количество японских автомобилей:", len(mpg_japan))

print("Средняя экономичность автомобилей США:", mpg_usa.mean())
print("Средняя экономичность автомобилей Японии:", mpg_japan.mean())

print("Дисперсия автомобилей США:", mpg_usa.var())
print("Дисперсия автомобилей Японии:", mpg_japan.var())

sns.boxplot(x="origin", y="mpg", data=df, order=["usa", "japan"])
plt.title("Топливная экономичность автомобилей США и Японии")
plt.xlabel("Регион происхождения")
plt.ylabel("Топливная экономичность")
plt.show()

t_statistic, p_value = stats.ttest_ind(
    mpg_usa, mpg_japan, equal_var=False
)

print("t-статистика:", t_statistic)
print("p-value:", p_value)

if p_value <= alpha:
    print("Нулевая гипотеза отвергается")
else:
    print("Нет оснований отвергать нулевую гипотезу")
```

**Вывод**

Японские автомобили в среднем экономичнее американских - они способны проехать большее расстояние на одинаковом количестве топлива. Т.к p-value меньше 0.05, различие между средними значениями является статистически значимым. Поэтому мы отвергаем нулевую гипотезу в пользу альтернативной

В качестве признаков используются:

Столбец name исключается из анализа, т.к содержит названия моделей автомобилей и имеет большое количество уникальных значений

```python
df_encoded = df.copy()

df_encoded = df_encoded.drop(columns=["name"])

df_encoded = pd.get_dummies(df_encoded, columns=["origin"], dtype=int)

display(df_encoded.head())

correlation_matrix = df_encoded.corr()

print("Матрица корреляций:")
display(correlation_matrix.round(3))

plt.figure(figsize=(12, 8))
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Матрица корреляций")
plt.show()

print("Корреляция признаков с целевым столбцом:")
display(correlation_matrix["mpg"].drop("mpg").sort_values().round(3))
```

**Вывод:**

Самая сильная отрицательная корреляция с топливной экономичностью видна у массы автомобиля, рабочего объема двигателя, мощности двигателя и количества цилиндров. Это значит, что тяжелые автомобили с более мощными и объемными двигателями обычно расходуют больше топлива

```python
matches = pd.read_csv("matches.csv")
teamstats = pd.read_csv("teamstats.csv")
participants = pd.read_csv("participants.csv")
stats1 = pd.read_csv("stats1.csv")
stats2 = pd.read_csv("stats2.csv")
champs = pd.read_csv("champs.csv")
teambans = pd.read_csv("teambans.csv")

stats_players = pd.concat([stats1, stats2], ignore_index=True)

del stats1, stats2

def analysis(data, numeric_columns, categorical_columns):
    print("РАЗМЕР ТАБЛИЦЫ:", data.shape)
    display(data.head())

    print("\nОБЩАЯ ИНФОРМАЦИЯ:")
    data.info()

    print("\nПРОПУСКИ:")
    display(data.isna().mean())

    if len(numeric_columns) > 0:
        numeric = data[numeric_columns]

        print("\nЧИСЛОВЫЕ ПЕРЕМЕННЫЕ:")

        statistics = pd.DataFrame({
            "Доля пропусков": numeric.isna().mean(),
            "Минимум": numeric.min(),
            "Максимум": numeric.max(),
            "Среднее": numeric.mean(),
            "Медиана": numeric.median(),
            "Дисперсия": numeric.var(),
            "Квантиль 0.1": numeric.quantile(0.1),
            "Квантиль 0.9": numeric.quantile(0.9),
            "Первый квартиль": numeric.quantile(0.25),
            "Третий квартиль": numeric.quantile(0.75)
        })

        display(statistics.round(3))

    if len(categorical_columns) > 0:
        categorical = data[categorical_columns]

        print("\nКАТЕГОРИАЛЬНЫЕ ПЕРЕМЕННЫЕ:")

        statistics_cat = pd.DataFrame({
            "Доля пропусков": categorical.isna().mean(),
            "Количество уникальных значений": categorical.nunique(),
            "Мода": categorical.mode().iloc[0]
        })

        display(statistics_cat)


# Матчи и команды
teamstats_sum = teamstats.groupby("matchid")[["towerkills", "inhibkills", "baronkills", "dragonkills", "harrykills"]].sum()

lol_matches = matches.merge(teamstats_sum, left_on="id", right_index=True)


# Игроки и чемпионы
lol_players = participants.merge(stats_players, on="id")

lol_players = lol_players.merge(champs.rename(columns={"id": "championid"}), on="championid", how="left")


# Баны и чемпионы
lol_bans = teambans.merge(champs.rename(columns={"id": "championid"}), on="championid", how="left")



print("\nАНАЛИЗ МАТЧЕЙ И КОМАНД:")

analysis(lol_matches,
    ["duration", "towerkills", "inhibkills",
     "baronkills", "dragonkills", "harrykills"],
    ["platformid", "queueid", "seasonid", "version"]
)

lol_players["wardsbought"] = pd.to_numeric(lol_players["wardsbought"], errors="coerce")

categorical_players = [
    "player", "championid", "name", "ss1", "ss2",
    "role", "position", "win", "item1", "item2",
    "item3", "item4", "item5", "item6", "trinket",
    "firstblood"
]

numeric_players = lol_players.drop(columns=["id", "matchid"] + categorical_players).select_dtypes(include="number").columns.tolist()

print("\nАНАЛИЗ ИГРОКОВ И ЧЕМПИОНОВ:")
analysis(lol_players, numeric_players, categorical_players)

print("\nСАМЫЕ ПОПУЛЯРНЫЕ ЧЕМПИОНЫ:")
display(lol_players["name"].value_counts().head(10))

print("\nСАМЫЕ ПОПУЛЯРНЫЕ ПОЗИЦИИ:")
display(lol_players["position"].value_counts())

print("\nАНАЛИЗ БАНОВ ЧЕМПИОНОВ:")

analysis(lol_bans, [], ["teamid", "championid", "banturn", "name"])

print("\nСАМЫЕ ЧАСТО ЗАБАНЕННЫЕ ЧЕМПИОНЫ:")
display(lol_bans["name"].value_counts().head(10))
```
